```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_excel("heart.xlsx")

# ===== FIRST FIVE RECORDS =====
print("===== FIRST FIVE RECORDS =====")
print(df.head())

# ===== SHAPE OF DATASET =====
print("\n===== SHAPE OF DATASET =====")
print(df.shape)

# ===== DATA TYPES =====
print("\n===== DATA TYPES =====")
print(df.dtypes)

# ===== DESCRIPTIVE STATISTICS =====
print("\n===== DESCRIPTIVE STATISTICS =====")
print(df.describe())

# ===== LINE CHART =====
age_target = df.groupby("age")["target"].mean()

plt.figure(figsize=(10,5))
plt.plot(age_target.index, age_target.values)
plt.title("Heart Disease Rate by Age")
plt.xlabel("Age")
plt.ylabel("Average Heart Disease Rate")
plt.show()

# ===== BAR CHART =====
sex_target = df.groupby("sex")["target"].mean()

plt.figure(figsize=(6,4))
plt.bar(sex_target.index.astype(str), sex_target.values)
plt.title("Heart Disease Rate by Gender")
plt.xlabel("Sex (0=Female, 1=Male)")
plt.ylabel("Average Heart Disease Rate")
plt.show()

# ===== PIE CHART =====
target_counts = df["target"].value_counts()

plt.figure(figsize=(7,7))
plt.pie(
    target_counts,
    labels=["Heart Disease","No Heart Disease"],
    autopct="%1.1f%%"
)
plt.title("Heart Disease Distribution")
plt.show()

# ===== HISTOGRAM =====
plt.figure(figsize=(8,5))
plt.hist(df["age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# ===== SCATTER PLOT =====
plt.figure(figsize=(8,5))
plt.scatter(df["age"], df["chol"])
plt.title("Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.show()

# ===== BOX PLOT =====
plt.figure(figsize=(6,5))
plt.boxplot(df["chol"])
plt.title("Box Plot of Cholesterol")
plt.ylabel("Cholesterol")
plt.show()

# ===== LINE COMPARISON CHART =====
age_data = df.groupby("age")[["chol","thalach"]].mean()

plt.figure(figsize=(10,5))
plt.plot(age_data.index, age_data["chol"], label="Cholesterol")
plt.plot(age_data.index, age_data["thalach"], label="Max Heart Rate")
plt.title("Cholesterol vs Max Heart Rate by Age")
plt.xlabel("Age")
plt.ylabel("Average Value")
plt.legend()
plt.show()
```
