```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_excel("heart.xlsx")

# 1. First Five Records
print("===== FIRST FIVE RECORDS =====")
print(df.head())

# 2. Shape of Dataset
print("\n===== SHAPE =====")
print(df.shape)

# 3. Data Types
print("\n===== DATA TYPES =====")
print(df.dtypes)

# 4. Filter Patients with Heart Disease
print("\n===== PATIENTS WITH HEART DISEASE =====")
print(df[df["target"] == 1].head())

# 5. Sort by Cholesterol (Descending)
print("\n===== SORT BY CHOLESTEROL =====")
print(df.sort_values(by="chol", ascending=False).head())

# 6. Statistical Analysis
print("\n===== AGE STATISTICS =====")
print("Mean:", df["age"].mean())
print("Median:", df["age"].median())
print("Mode:", df["age"].mode()[0])
print("Standard Deviation:", df["age"].std())

# 7. Descriptive Statistics
print("\n===== DESCRIPTIVE STATISTICS =====")
print(df.describe())

# 8. Grouping and Aggregation
print("\n===== AVERAGE CHOLESTEROL BY TARGET =====")
print(df.groupby("target")["chol"].mean())

# 9. Correlation Matrix
print("\n===== CORRELATION MATRIX =====")
corr = df.corr(numeric_only=True)
print(corr)

# 10. Bar Chart
df.groupby("target")["chol"].mean().plot(kind="bar")
plt.title("Average Cholesterol by Heart Disease")
plt.xlabel("Target (0=No Disease, 1=Disease)")
plt.ylabel("Average Cholesterol")
plt.show()

# 11. Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(df["age"], df["chol"])
plt.title("Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.show()

# 12. Histogram
plt.figure(figsize=(8,5))
plt.hist(df["age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
